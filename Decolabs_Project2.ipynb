```python
def get_expense(entry):
    """Convert text to a valid positive number, or return None if invalid."""
    try:
        amount = float(entry)
    except ValueError:
        return None
    if amount <= 0:
        return None
    return amount


def main():
    total = 0.0
    count = 0

    print("=== Expense Tracker ===")
    print("Enter each expense. Type 'quit' to finish.\n")

    while True:
        entry = input("Enter expense (or 'quit'): ").strip()

        if entry.lower() == "quit":
            break

        amount = get_expense(entry)
        if amount is None:
            print("Invalid data. Enter a positive number.")
            continue

        total += amount
        count += 1
        print(f"Added {amount:.2f}. Running total: {total:.2f}")

    print("\n------------------------")
    print(f"Transactions: {count}")
    print(f"FINAL TOTAL: ${total:.2f}")


if __name__ == "__main__":
    main()
```
