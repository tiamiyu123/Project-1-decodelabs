```python
import math
import secrets
import string

MIN_LENGTH = 8


def get_length():
    """Keep asking until the user enters a valid length."""
    while True:
        entry = input(f"Enter password length (minimum {MIN_LENGTH}): ").strip()
        try:
            length = int(entry)
        except ValueError:
            print("Please enter a whole number.")
            continue
        if length < MIN_LENGTH:
            print(f"Length must be at least {MIN_LENGTH}.")
            continue
        return length


def generate_password(length, use_symbols=False):
    """Return a secure random password with at least one of each character type."""
    pool = string.ascii_letters + string.digits
    required = [
        secrets.choice(string.ascii_lowercase),
        secrets.choice(string.ascii_uppercase),
        secrets.choice(string.digits),
    ]
    if use_symbols:
        pool += string.punctuation
        required.append(secrets.choice(string.punctuation))

    remaining = [secrets.choice(pool) for _ in range(length - len(required))]
    chars = required + remaining
    secrets.SystemRandom().shuffle(chars)
    return "".join(chars), len(pool)


def entropy_bits(length, pool_size):
    return length * math.log2(pool_size)


def main():
    print("=== Secure Password Generator ===")
    length = get_length()
    symbols = input("Include symbols like @#$? (y/n): ").strip().lower() == "y"

    password, pool_size = generate_password(length, symbols)

    print(f"\nYour password: {password}")
    print(f"Length: {length} | Character pool: {pool_size}")
    print(f"Estimated entropy: {entropy_bits(length, pool_size):.1f} bits")


if __name__ == "__main__":
    main()
```
