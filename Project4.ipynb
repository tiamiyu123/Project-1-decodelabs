```python
QUESTIONS = [
    {"question": "What is the capital of France?", "answer": "paris"},
    {"question": "Which planet has famous rings?", "answer": "saturn"},
    {"question": "What is 7 x 8?", "answer": "56"},
]


def ask_question(question, correct_answer):
    """Ask one question. Return True if the answer is correct."""
    guess = input(f"{question} ").strip().lower()
    if guess == correct_answer:
        print("Correct! +1 point\n")
        return True
    else:
        print(f"Wrong. The answer was {correct_answer}.\n")
        return False


def main():
    score = 0                                   # set ONCE, before the loop
    total = len(QUESTIONS)

    print("=== General Knowledge Quiz ===\n")
    for number, q in enumerate(QUESTIONS, start=1):
        print(f"Question {number} of {total}")
        if ask_question(q["question"], q["answer"]):
            score += 1

    print(f"Final score: {score:>2} / {total}")


if __name__ == "__main__":
    main()
```
