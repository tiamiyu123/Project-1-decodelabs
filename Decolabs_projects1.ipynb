```python
import json
import os

FILE = "tasks.json"


def load_tasks():
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            return json.load(f)
    return []


def save_tasks(tasks):
    with open(FILE, "w") as f:
        json.dump(tasks, f, indent=2)


def add_task(tasks, task):
    tasks.append(task)


def view_tasks(tasks):
    if not tasks:
        print("No tasks yet.")
        return
    for i, task in enumerate(tasks, start=1):
        print(f"{i}. {task}")


def main():
    my_tasks = load_tasks()

    while True:
        print("\n1. Add task")
        print("2. View tasks")
        print("3. Quit")
        choice = input("Choose: ").strip()

        if choice == "1":
            task = input("Enter task: ").strip()
            if task:
                add_task(my_tasks, task)
                save_tasks(my_tasks)
                print("Task added!")
            else:
                print("Task cannot be empty.")
        elif choice == "2":
            view_tasks(my_tasks)
        elif choice == "3":
            print("Goodbye!")
            break
        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
